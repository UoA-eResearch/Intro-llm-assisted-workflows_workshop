# API call notebook (for reference)

## 1. Set up


```python
from openai import OpenAI
from IPython.display import Markdown
from dotenv import load_dotenv
import os

load_dotenv() # Load environment variables from .env file


api_key = os.getenv("api_key")
base_url = os.getenv("base_url")

model_name = "Qwen/Qwen3.6-27B"# Can get this from lais dashboard or compute provider
```

```python
question = "What is the capital of France?" # this is your prompt

lais_01 = OpenAI(api_key=api_key, base_url=base_url) # this is our main call

messages = [{"role": "user", "content": question}] # This is how prompts are structured

response = lais_01.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content
display(Markdown(answer))
```
